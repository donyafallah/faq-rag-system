```python
from llama_index.core import Document
```

```python
documents = []

for index, row in df.iterrows():

    text = f"""
    Category: {row['category']}
    Question: {row['question']}
    Answer: {row['answer']}
    """

    document = Document(
        text = text,
        metadata =  {
            'category' : row["category"],
            "source": "faq_dataset"
        }
    )

    documents.append(document)
```

```python
print("Total Documents:", len(documents))

print("\nFirst Document:")
print(documents[0].text)

print("\nFirst Document Metadata:")
print(documents[0].metadata)
```

```python
import torch
from llama_index.embeddings.huggingface import HuggingFaceEmbedding


device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)
print("CUDA available:", torch.cuda.is_available())
```

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.core import Settings

embed_model = HuggingFaceEmbedding(
    model_name="intfloat/multilingual-e5-small",
    device="cpu",
    query_instruction="query: ",
    text_instruction="passage: "
)

Settings.embed_model = embed_model

print("Embedding model loaded successfully!")
```

```python
from llama_index.vector_stores.chroma import ChromaVectorStore
from llama_index.core import StorageContext
import chromadb
```

```python
db = chromadb.PersistentClient(path="../chroma_db")

chroma_collection = db.get_or_create_collection(
    name="faq_collection",
    metadata={"hnsw:space": "cosine"}
)

vector_store = ChromaVectorStore(
    chroma_collection=chroma_collection
)

storage_context = StorageContext.from_defaults(
    vector_store=vector_store
)

print("ChromaDB connected successfully!")
print("Existing records:", chroma_collection.count())
```

```python
from llama_index.core import VectorStoreIndex

index = VectorStoreIndex.from_documents(
    documents,
    storage_context=storage_context,
    embed_model=embed_model,
    show_progress=True
)

print("Vector Index created successfully!")
print("Total records:", chroma_collection.count())
```

```python
print("Number of documents:", len(documents))
print("Number of Chroma records:", chroma_collection.count())

stored_data = chroma_collection.get(
    include=["metadatas"]
)

print("Stored records:", len(stored_data["ids"]))
print("First 10 IDs:", stored_data["ids"][:10])
print("First 10 metadata:", stored_data["metadatas"][:10])
```
