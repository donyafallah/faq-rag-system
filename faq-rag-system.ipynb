```python
import pandas as pd
```

```python
df = pd.read_csv("faq_qa_pairs.csv")
print(df.head())
```

```python
print("Shape: ", df.shape)
print("Column: ", df.columns)
print("Null Column:\n", df.isnull().sum())
```

```python
print(df["category"].value_counts(), "\n")
print(df['question'].value_counts(), "\n")
print(df["answer"].value_counts(), "\n")
```

```python
row = df.iloc[2]
row
```

```python
documents = []

for index, row in df.iterrows():
    document = f""" 
       Category : {row['category']}
       Question : {row['question']}
       Answer : {row['answer']}
    """
    documents.append(document)

print(documents[0])
print(documents[1])
```

```python
document_lengths = []

for doc in documents:
    temp = len(doc)
    document_lengths.append(temp)

# documents_length = [len(doc) for doc in documents]

print("Number of documents: ", len(documents))
print("Shortest document: ", min(document_lengths))
print("Longest document: ", max(document_lengths))
print("Average document length: ", sum(document_lengths)/ len(document_lengths))
```

```python
from openai import OpenAI
from getpass import getpass

api_key = getpass("Enter your API key: ")

client = OpenAI(
    api_key=api_key,
    base_url="https://api.avalai.ir/v1"
)
```

```python
response = client.embeddings.create(
    model="text-embedding-3-small",
    input=documents
)
```

```python
embeddings = []

sorted_data = sorted(
    response.data,
    key=lambda x: x.index
)

for item in sorted_data:
    embeddings.append(item.embedding)

print("Number of Documents: ", len(embeddings))
print("Dimension: ", len(embeddings[0]))
```

```python
import chromadb
```

```python
chroma_client = chromadb.PersistentClient(
    path="./chromadb-faq-rag-system"
)
```

```python
collection = chroma_client.get_or_create_collection(
    name="faq_collection",
    metadata={"hnsw:space":"cosine"}
)
```

```python
ids = []
metadatas = []

for i in range(len(documents)):
    ids.append(f"id_{i}")

    metadata = {
        "Category" : df.iloc[i]["category"],
        "Question" : df.iloc[i]["question"],
        "Answer" : df.iloc[i]["answer"]
    }

    metadatas.append(metadata)
```

```python
collection.add(
    ids=ids,
    embeddings=embeddings,
    metadatas=metadatas,
    documents=documents
)
```

```python
print("Total Documents:", collection.count())
```

```python
query = "چه نوع حمایت‌هایی برای شرکت‌های دانش‌بنیان در نظر گرفته شده است؟"

query_response = client.embeddings.create(
    model="text-embedding-3-small",
    input=query
)

query_embedding = query_response.data[0].embedding
```

```python
print("Query Response Appearance: ", query_response)
print("Embedding Dimensional: ", len(query_embedding))
```

```python
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=5
)

for i in range(len(results["ids"][0])):
    print("Result:", i+1)
    print("Document:", results["documents"][0][i])
    print("Distance:", results["distances"][0][i])
```
